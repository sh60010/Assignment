```python
# Question 9  

Create a dictionary of five products.

Example

    Laptop
    Mouse
    Keyboard
    Monitor
    Printer

Each product should contain

    Product Name
    Price
    Quantity

Write a program to

    Display all products.
    Print the product with the highest price.
    Calculate the total inventory value

(Formula:  Inventory Value = Price × Quantity)
```

```python
products = {
    "Laptop": {
        "Product Name": "Laptop",
        "Price": 50000,
        "Quantity": 10
    },
    "Mouse": {
        "Product Name": "Mouse",
        "Price": 500,
        "Quantity": 50
    },
    "Keyboard": {
        "Product Name": "Keyboard",
        "Price": 1200,
        "Quantity": 30
    },
    "Monitor": {
        "Product Name": "Monitor",
        "Price": 15000,
        "Quantity": 8
    },
    "Printer": {
        "Product Name": "Printer",
        "Price": 10000,
        "Quantity": 5
    }
}

for product, details in products.items():
    print(product)
    print("  Product Name:", details["Product Name"])
    print("  Price:", details["Price"])
    print("  Quantity:", details["Quantity"])
    print()

highest_price = 0
highest_product = ""

for product, details in products.items():
    if details["Price"] > highest_price:
        highest_price = details["Price"]
        highest_product = product
print(highest_product, "-", highest_price)
total_inventory = 0
for details in products.values():
    total_inventory += details["Price"] * details["Quantity"]

print("Total Inventory Value:", total_inventory)
```

```python
# Question 10

Create a dictionary of five students and their marks.

Example

    Alice : 85
    Bob : 92
    Charlie : 78
    David : 88
    Emma : 95

Write a Python program to

    Find the student with the highest marks.
    Find the student with the lowest marks.
    Calculate the average marks.
    Display students who scored more than 85.
```

```python
# Create a dictionary of students and their marks
students = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 88,
    "Emma": 95
}

# 1. Find the student with the highest marks
highest_student = max(students, key=students.get)
print("Student with Highest Marks:")
print(highest_student, "-", students[highest_student])

# 2. Find the student with the lowest marks
lowest_student = min(students, key=students.get)
print("\nStudent with Lowest Marks:")
print(lowest_student, "-", students[lowest_student])

# 3. Calculate the average marks
average = sum(students.values()) / len(students)
print("\nAverage Marks:", average)

# 4. Display students who scored more than 85
print("\nStudents who scored more than 85:")
for student, marks in students.items():
    if marks > 85:
        print(student, "-", marks)
```
